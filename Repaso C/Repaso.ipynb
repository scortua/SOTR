## Tipos Enteros  
### *CHAR* 
De 8 bits o 1 byte  
* Almacena caracteres y cadenas en código ASCII.  
* Ocupa 1 byte de memoria.  
* Puede ser signed o unsigned, afectando el rango de valores.

### *SHORT*
De 16 bits o 2 bytes  
* Almacena valores enteros de -32,768 a 32,767 (signed) o 0 a 65,535 (unsigned).  
* Ocupa 2 bytes de memoria.  
* Útil para ahorrar espacio en variables con rango limitado.

### *INT*
Depende del microcontrolador o procesador, generalmente 4 bytes (32 bits).  
* Rango típico: -2,147,483,648 a 2,147,483,647 (signed).  
* Ocupa 4 bytes de memoria.  
* Es el tipo entero más usado por su equilibrio entre rango y eficiencia.

### *LONG*
De 4 a 8 bytes según la arquitectura.  
* Rango mayor que int.  
* Ocupa más memoria, útil para valores grandes.

### *LONGLONG*
De 8 bytes (64 bits).  
* Rango extremadamente grande para enteros.  
* Ocupa 8 bytes de memoria.

## Tipos Flotantes

### *FLOAT*
De 4 bytes  
* Guarda números de punto flotante con precisión simple (~7 decimales).  
* Ocupa 4 bytes de memoria.  
* Útil para cálculos rápidos y menos precisos.

### *DOUBLE*
De 8 bytes  
* Guarda números de punto flotante con doble precisión (~15 decimales).  
* Ocupa 8 bytes de memoria.  
* Usado cuando se requiere mayor precisión en cálculos.

1 (SIGNO) - 23 (MANTISA) - 8 (EXPONENTE) - IEEE 754

### *VOID*
No almacena datos, representa ausencia de tipo.  
* Usado para funciones que no retornan valor o para punteros genéricos.  
* No ocupa memoria directamente.

### *STRUCT*
Agrupa diferentes tipos de datos bajo una sola entidad.  
* Cada miembro ocupa memoria según su tipo y alineación.  
* Útil para modelar objetos complejos.

### *UNION*
Permite almacenar diferentes tipos en el mismo espacio de memoria.  
* Todos los miembros comparten la misma dirección.  
* Ocupa el tamaño del miembro más grande.

### *ENUM*
Define conjuntos de constantes enteras con nombres.  
* Ocupa el mismo espacio que un int.  
* Facilita la legibilidad y el mantenimiento del código.

## Tipos de Datos

* __signed__: Permite que una variable almacene valores positivos y negativos. Por defecto, los tipos numéricos son signed.
* __unsigned__: Solo almacena valores positivos, duplicando el rango superior respecto a signed.
* __static__: Mantiene el valor de la variable entre llamadas a funciones. Limita el alcance a su archivo o función.
* __const__: Indica que el valor de la variable no puede cambiar después de su inicialización.
* __auto__: Especifica almacenamiento automático (por defecto en variables locales). Suele omitirse porque es implícito.
* __* (puntero)__: Permite almacenar la dirección de memoria de otra variable, facilitando el manejo dinámico de datos.
* __volatile__: Indica que la variable puede cambiar en cualquier momento fuera del control del programa (por ejemplo, por hardware).
* __extern__: Declara una variable definida en otro archivo, permitiendo su uso en múltiples archivos fuente.

## Instrucciones

Las instrucciones que sigue una Unidad Aritmético Lógica (ALU) permiten realizar operaciones matemáticas y lógicas sobre los datos procesados por el microprocesador. A continuación se describen las principales:

* **+** : Suma dos operandos.
* **-** : Resta el segundo operando al primero.
* **\*** : Multiplica dos operandos.
* **/** : Divide el primer operando por el segundo.
* **>** : Compara si el primer operando es mayor que el segundo.
* **<** : Compara si el primer operando es menor que el segundo.
* **==** : Verifica si ambos operandos son iguales.
* **!=** : Verifica si los operandos son diferentes.
* **AND** : Realiza una operación lógica AND bit a bit entre dos operandos.
* **OR** : Realiza una operación lógica OR bit a bit entre dos operandos.
* **XOR** : Realiza una operación lógica XOR bit a bit entre dos operandos.
* **NOT** : Realiza una operación lógica NOT (inversión de bits) sobre un operando.
* **>>** : Desplaza los bits del operando a la derecha (shift right).
* **<<** : Desplaza los bits del operando a la izquierda (shift left).

Estas instrucciones permiten a la ALU ejecutar cálculos, tomar decisiones y manipular datos a nivel de bits, siendo fundamentales para el funcionamiento de cualquier procesador.

## Arquitectura Procesador
La arquitectura de procesadores define cómo se organizan y conectan los componentes internos para ejecutar instrucciones. Incluye el diseño de la Unidad Central de Procesamiento (CPU), la memoria, los buses y los registros. Las arquitecturas más comunes son:

* __RISC__ (Reduced Instruction Set Computer): Utiliza un conjunto reducido y simple de instrucciones, permitiendo una ejecución rápida y eficiente. Cada instrucción suele realizar una sola operación y se ejecuta en un solo ciclo de reloj. Ejemplo: ARM, MIPS.
* __CISC__ (Complex Instruction Set Computer): Ofrece un conjunto amplio y complejo de instrucciones, algunas capaces de realizar tareas avanzadas en una sola instrucción. Esto puede simplificar la programación pero aumentar la complejidad del hardware. Ejemplo: x86, Intel 8086.

## Manejo de memoria

* __HARVARD__:  
    La arquitectura Harvard separa físicamente la memoria y los buses para instrucciones y datos. Esto permite que el procesador acceda simultáneamente a instrucciones y datos, aumentando la velocidad de procesamiento. El manejo de datos es más eficiente, ya que evita conflictos entre la lectura/escritura de datos y la obtención de instrucciones. Es común en microcontroladores y sistemas embebidos.

* __VON NEUMAN__:  
    En la arquitectura Von Neumann, instrucciones y datos comparten la misma memoria y el mismo bus. Esto simplifica el diseño pero puede causar cuellos de botella, ya que el procesador no puede acceder a instrucciones y datos al mismo tiempo. El manejo de datos es menos eficiente en comparación con Harvard, pero es más flexible y ampliamente utilizado en computadoras personales y servidores.

Se puede decir que lo que hace un microprocesador es manejar datos, **moverlos y procesarlos**.

# Sentencia de Control

Las sentencias de control en C permiten modificar el flujo de ejecución de un programa, tomando decisiones, repitiendo acciones o seleccionando entre múltiples alternativas. Son fundamentales para crear programas dinámicos y adaptativos, ya que permiten responder a diferentes condiciones y situaciones durante la ejecución.

Las principales sentencias de control en C son:

- `if` / `else`
- `switch`
- `for`
- `while`
- `do-while`
- `break`
- `continue`
- `goto`

Estas sentencias permiten estructurar la lógica de los programas y son esenciales para resolver problemas complejos de manera eficiente.

## IF
La sentencia `if` permite ejecutar un bloque de código solo si se cumple una condición específica. Es fundamental para la toma de decisiones en los programas, ya que permite controlar el flujo de ejecución según el resultado de una expresión lógica. Si la condición es verdadera, se ejecutan las instrucciones dentro del bloque; si es falsa, se omiten.
```c
if (condition)  // Only true or different of 0
{
    instrution
}
else            // Only false or equal of 0
{
    instrution
}
```
* 0 $\rightarrow$ False
* 1 $\rightarrow$ True

Puede verse que cualquier cosa que sea **diferente de 0** entra a la sentencia. Si no se cumple se puede usar el else para ejecutar lo necesario.

## SWITCH
La sentencia `switch` permite seleccionar entre múltiples opciones según el valor de una variable o expresión. Es útil cuando se tienen varios casos posibles y se desea ejecutar diferentes bloques de código para cada uno.
```c
switch (variable) {
    case valor1:
        // instrucciones para valor1
        break;
    case valor2:
        // instrucciones para valor2
        break;
    default:
        // instrucciones si no coincide ningún caso
        break;
}
```
Cada `case` compara el valor de la variable con un valor específico. Si hay coincidencia, se ejecutan las instrucciones correspondientes. El `break` evita que se ejecuten los siguientes casos. El bloque `default` se ejecuta si ningún caso coincide.

Solamente son validos valores enteros en el switch. Lo normal es que se ejecuta un codigo en cierto valor, pero no es obligatorio colocar un _break_ puede saltar al siguiente codigo de switch hasta terminar el switch o encontrar un break.

## WHILE
La sentencia `while` en C permite ejecutar repetidamente un bloque de código mientras se cumpla una condición lógica. Es ideal para situaciones donde no se conoce de antemano cuántas veces debe repetirse el ciclo, y la condición se evalúa antes de cada iteración.
```c
While (verdadero)
{
    Instruccion
}
```
### Break
Tambien se puede agregar un `break` a la instruccion para romper la ejecucion en alguna parte del bloque de codigo.
```c
While (verdadero)
{
    Instruccion
    break
    Instruccion no ejecutada
}
```
### Continue
Tambien se puee agregar un `continue` para seguir con la continuidad a la ejecucion del while.

La condición se verifica al final de cada iteración, garantizando que el bloque se ejecute al menos una vez. Es comúnmente utilizada para menús interactivos, validaciones o procesos que requieren una ejecución inicial obligatoria.
### Break
El `break` en un ciclo `do-while` permite salir inmediatamente del ciclo, sin esperar a que la condición se evalúe al final. Es útil cuando se cumple una condición específica dentro del bloque y se desea terminar la repetición antes de tiempo.

Por ejemplo:
```c
do {
    // instrucciones
    if (condición_de_salida) {
        break;
    }
    // más instrucciones
} while (condición);
```
Al ejecutarse el `break`, el flujo de ejecución continúa con la siguiente instrucción después del ciclo `do-while`.
### Continue
El `continue` en un ciclo `do-while` permite saltar el resto de las instrucciones del bloque actual y pasar directamente a la evaluación de la condición para la siguiente iteración. Es útil cuando se desea omitir parte del código bajo ciertas circunstancias, pero continuar repitiendo el ciclo.

Por ejemplo:
```c
do {
    // instrucciones
    if (condición_de_salto) {
        continue;
    }
    // más instrucciones
} while (condición);
```
Cuando se ejecuta el `continue`, las instrucciones posteriores en el bloque se omiten y se verifica la condición para decidir si el ciclo continúa.

## FOR
El ciclo `for` en C permite ejecutar un bloque de código un número determinado de veces, controlando la inicialización, condición y actualización de una variable en una sola línea. Es ideal para iteraciones conocidas, como recorrer arreglos o repetir acciones un número específico de veces.

Su sintaxis es:
```c
for (inicialización; condición; actualización) 
{
    // instrucciones
}
```
- **Inicialización:** Se ejecuta una vez al inicio del ciclo.
- **Condición:** Se evalúa antes de cada iteración; si es verdadera, se ejecuta el bloque.
- **Actualización:** Se ejecuta al final de cada iteración.

Ejemplo:
```c
for (int i = 0; i < 10; i++) 
{
    printf("%d\n", i);
}
```
En este ejemplo, la variable `i` inicia en 0 y el ciclo se repite mientras `i` sea menor que 10, incrementando `i` en cada iteración.
```c
for (x=0,i=0,y=funcion() ; x+y*i=funcion() ; i++,x--)
{
    Instruccion
}
```
### Break
El `break` en un ciclo `for` permite salir inmediatamente del ciclo, sin esperar a que la condición se vuelva falsa. Es útil cuando se cumple una condición específica dentro del bloque y se desea terminar la repetición antes de tiempo.

Por ejemplo:
```c
for (int i = 0; i < 10; i++) {
    if (i == 5) {
        break;
    }
    printf("%d\n", i);
}
```
En este caso, el ciclo se detiene cuando `i` es igual a 5, por lo que solo se imprimen los valores del 0 al 4.

### Continue
El `continue` en un ciclo `for` permite saltar el resto de las instrucciones del bloque actual y pasar directamente a la actualización de la variable y la siguiente iteración. Es útil cuando se desea omitir parte del código bajo ciertas circunstancias, pero continuar repitiendo el ciclo.

Por ejemplo:
```c
for (int i = 0; i < 10; i++) {
    if (i % 2 == 0) {
        continue;
    }
    printf("%d\n", i);
}
```
En este caso, solo se imprimen los valores impares del 0 al 9, ya que cuando `i` es par, el `continue` salta la instrucción `printf`.

## Definiciones h y c

### .h
Los archivos `.h` (header o encabezado) en C se utilizan para declarar funciones, tipos de datos, macros y variables globales que serán compartidas entre varios archivos fuente (`.c`). Permiten separar la interfaz de la implementación, facilitando la modularidad y el mantenimiento del código. En un archivo `.h` se incluyen las declaraciones, mientras que la lógica se implementa en el archivo `.c` correspondiente. Al incluir un `.h` con `#include`, otros archivos pueden acceder a sus definiciones sin duplicar código. Esto es esencial en proyectos grandes y colaborativos.

#### `#include`
La directiva `#include` en C permite incorporar el contenido de un archivo externo (generalmente `.h`) en el archivo fuente actual antes de la compilación. Esto facilita el uso de funciones, macros y definiciones declaradas en otros archivos, promoviendo la reutilización y organización del código.

- `#include "archivo.h"`: Busca el archivo en el directorio del proyecto primero y luego en las rutas estándar.
- `#include <archivo.h>`: Busca el archivo únicamente en las rutas estándar del compilador (usado para librerías del sistema).

El uso correcto de `#include` evita duplicidad de código y errores de compilación, asegurando que todas las declaraciones necesarias estén disponibles en cada archivo fuente que las requiera.

#### `#define`
La directiva `#define` en C se utiliza para crear macros, es decir, constantes simbólicas o fragmentos de código que el preprocesador reemplaza antes de la compilación. Permite definir valores constantes, expresiones o funciones simples que se reutilizan en todo el programa, mejorando la legibilidad y facilitando cambios globales.

- `#define PI 3.1416`  
    Define una constante llamada `PI` con el valor `3.1416`.

- `#define MAX(a, b) ((a) > (b) ? (a) : (b))`  
    Define una macro para obtener el valor máximo entre dos variables.

El uso de `#define` es común para evitar el uso de números mágicos, facilitar configuraciones y crear funciones simples sin sobrecarga de llamada. Sin embargo, hay que tener cuidado con los paréntesis y el alcance, ya que el reemplazo es textual y puede causar errores si no se usa correctamente.

El inicio `#` se usa para uso del directivo en precompilacion.

```c
#include "" o <> // Inclucion
#define 
Definicion de tipos
Definicion de funciones
```

#### .c
Los archivos `.c` en C contienen la implementación del código fuente, es decir, la lógica de las funciones, procedimientos y estructuras declaradas en los archivos de encabezado `.h`. En estos archivos se desarrollan las operaciones, algoritmos y el flujo principal del programa. Cada archivo `.c` puede incluir uno o varios encabezados mediante la directiva `#include`, permitiendo el acceso a las declaraciones necesarias. Esta separación entre declaración (.h) e implementación (.c) mejora la organización, facilita la reutilización y el trabajo colaborativo, ya que cada módulo puede desarrollarse y mantenerse de forma independiente. Al compilar, todos los archivos `.c` se combinan para formar el programa ejecutable.


# Operadores

### (.) Structure Member Operator
```c
name.member
```
Sirve para acceder a un campo de una estructura cuando tienes la variable directamente (no puntero). Es decir **Dame el miembro x de la estructura**.
```c
typedef struct {
    int id;
    float value;
} my_data_t;

my_data_t d1;   // variable de tipo struct
d1.id = 10;     // uso el operador .
d1.value = 3.14;
```
* d1 es la variable
* d1.id se accede al campo id de d1
* d1.value se accede al campo value de d1

### (->) Structure Pointer
```c
pointer->member
```
El operador `->` sirve también para trabajar con punteros de funciones. Es decir que funciona para acceder a un miembro de una estructura a tráves de un puntero. Como (*puntero).campo
```c
typedef struct {
    int id;
    float value;
} my_data_t;

my_data_t d = { .id = 1, .value = 3.14 };
my_data_t* p = &d;   // p es un puntero a d

// Acceso con ->
printf("%d\n", p->id);      // imprime 1
printf("%.2f\n", p->value); // imprime 3.14
// lo mismo que:
printf("%d\n", (*p).id);
printf("%.2f\n", (*p).value);
```

- Operadores: ++ (incremento) y -- (decremento).  
- Formas:
    - Prefijo: ++a / --a — incrementa/decrementa primero y devuelve el nuevo valor.
    - Sufijo: a++ / a-- — devuelve el valor original y luego modifica la variable.
- Aplicables a lvalues numéricos y punteros (en punteros avanza/retrocede por el tamaño del tipo).

Ejemplos:
```c
int a = 5;
int b = ++a; // prefijo: a == 6, b == 6

int c = 5;
int d = c++; // sufijo: c == 6, d == 5

int arr[3] = {10,20,30};
int *p = arr;
p++;         // p ahora apunta a arr[1]
```

### Maths Operations
```c
+ - ! ~
```
- `+` — operador binario de suma (`a + b`) y operador unario que deja el valor igual (`+x`). Usado para sumar aritméticamente valores numéricos.

- `-` — operador binario de resta (`a - b`) y operador unario de negación (`-x`) que invierte el signo del operando.

- `!` — operador lógico NOT (`!expr`): devuelve `1` si `expr` es 0, y `0` si `expr` es distinto de 0. Se usa para negar condiciones.

- `~` — operador de complemento a nivel de bits (`~x`): invierte todos los bits del operando entero. En representación Two's complement, `~x` equivale a `-x - 1`.

### (*pointer &name) Indirection Via Pointer, Address Of Object
```c
*x &x
```
- & — operador "address-of": obtiene la dirección en memoria de una variable. El resultado es un puntero cuyo tipo depende del tipo de la variable.
- * — operador de indireccion (dereference): accede al objeto al que apunta un puntero. También se usa en declaraciones para indicar que una variable es un puntero.

Ejemplos prácticos:
```c
int x = 5;        // x es una variable entera
int *p = &x;      // p guarda la dirección de x (tipo: int*)
printf("%d\n", *p); // imprime 5 (se accede al valor apuntado)

*p = 10;          // modifica el valor de x vía el puntero
printf("%d\n", x); // ahora imprime 10
```

Notas importantes:
- El tipo de &x es "puntero a tipo de x" (por ejemplo, int -> int*). El tipo de *p es "tipo apuntado" y puede usarse como lvalue para asignar.
- No desreferenciar punteros no inicializados o NULL: causa comportamiento indefinido.
```c
int *q = NULL;
// *q = 1; // peligro: UB (posible fallo en tiempo de ejecución)
```
- Los punteros pueden apuntar a arrays, estructuras o funciones; la aritmética de punteros avanza según el tamaño del tipo apuntado.
- Declaración vs uso: int *p; declara un puntero, p = &x; lo inicializa; *p lee o escribe el valor en esa dirección.

Ejemplo de puntero a puntero:
```c
int a = 3;
int *pa = &a;
int **ppa = &pa;
printf("%d\n", **ppa); // imprime 3
```

### ((type)expr) Cast Expression to Type
```c
(type)expr
```
Trata esta expresión (expr) como si fuera de este otro tipo (type)
```c
int a = 5, b = 2;
float r;

r = (float)a / b;   // convierto 'a' a float antes de dividir
printf("%f\n", r);
```

### (sizeof) Size of an Object
```c
sizeof(type)
```
Devuelve el **tamaño en bytes** que ocupa en memoria una variable, un tipo o una expresión.
- Se evalúa **en tiempo de compilación** (no en ejecución).
- El valor que devuelve es de tipo `size_t` (un entero sin signo).
```c
sizeof(char)   = 1
sizeof(int)    = 4
sizeof(float)  = 4
sizeof(double) = 8
```

### multiply divide modulus
Los operadores aritméticos de multiplicación, división y módulo en C se usan para operaciones básicas entre operandos numéricos y tienen comportamiento distinto según el tipo (entero o punto flotante), además de reglas sobre precedencia y asociatividad.

- Precedencia y asociatividad:
    - *, / y % comparten la misma precedencia y se evalúan de izquierda a derecha (asociatividad izquierda).
    - Se ejecutan antes que + y -.

- Multiplicación (*)  
    - Multiplica dos operandos. Funciona con enteros y con punto flotante.
    - Ejemplo:
        ```c
        int a = 4 * 3;       // 12
        double b = 2.5 * 4;  // 10.0
        ```

- División (/)  
    - Si ambos operandos son enteros, se realiza división entera (el resultado truncado hacia cero desde C99). Si alguno es de punto flotante, se realiza división en punto flotante.
    - Evitar dividir por cero: provoca comportamiento indefinido para enteros; para flotantes genera infinito/NaN según IEEE.
    - Ejemplo:
        ```c
        int x = 7 / 3;       // 2  (trunca hacia 0)
        int y = -7 / 3;      // -2 (trunca hacia 0)
        double z = 7.0 / 3;  // 2.333333...
        ```

- Módulo (%)  
    - Solo definido para operandos enteros (no para floats). Devuelve el resto de la división entera: a % b == a - (a / b) * b.
    - Desde C99, el cociente se trunca hacia cero, por lo que el signo del resto coincide con el del dividendo (primer operando).
    - Ejemplo:
        ```c
        int r1 = 7 % 3;   // 1
        int r2 = -7 % 3;  // -1  (signo igual al del dividendo)
        ```

### << >> left, right shift bit ops

Los operadores de desplazamiento en C mueven los bits de un operando hacia la izquierda o hacia la derecha.

- _<<_ (left shift)
    - Desplaza los bits hacia la izquierda n posiciones. Cada desplazamiento equivale a multiplicar por 2 (para enteros sin signo) si no hay overflow.
    - Los bits de la derecha se rellenan con ceros.
    - Para tipos signed, desplazar hacia la izquierda y provocar que bits significativos se pierdan puede producir comportamiento indefinido. Es más seguro usar tipos unsigned para manipulación de bits.
    - Ejemplo: multiplicar por potencias de dos:
        ```c
        uint32_t x = 3;
        uint32_t y = x << 2; // y == 12 (3 * 2^2)
        ```

- _>>_ (right shift)
    - Desplaza los bits hacia la derecha n posiciones.
    - Para tipos unsigned el comportamiento es lógico: se rellenan ceros por la izquierda.
    - Para tipos signed el comportamiento es implementation-defined: puede ser lógico (ceros) o aritmético (se replica el bit de signo), habitualmente los compiladores realizan shift aritmético para conservar el signo.
    - Ejemplo: dividir por potencias de dos (sin redondeo hacia infinito negativo si es signed):
        ```c
        uint32_t a = 20;
        uint32_t b = a >> 2; // b == 5
        ```

### Conditional Expression
```c
expr1 ? expr2 : expr3
```
Es un operador ternario que selecciona entre dos expresiones según una condición: evalúa primero la condición; si es distinta de 0 se evalúa y devuelve la segunda expresión, si es 0 se evalúa y devuelve la tercera. Solo se evalúa una de las dos ramas (comportamiento tipo short‑circuit), por lo que los efectos secundarios solo ocurren en la rama ejecutada.

Puntos clave:
- Evaluación: condición → una sola rama → resultado (la expresión completa tiene valor).
- Tipos: las dos ramas deben ser de tipos compatibles; se aplican las conversiones usuales (promociones/conversiones aritméticas o conversión de punteros cuando procede).
- L‑value: la expresión condicional puede ser l‑value solo si ambas ramas lo son y tienen tipos compatibles.
- Asociatividad: es asociativo por la derecha (permite anidar sin paréntesis, pero usar paréntesis mejora la legibilidad).
- Claridad: útil para asignaciones o retornos sencillos; no es recomendable para lógica compleja por razones de legibilidad.
- Sintaxis obligatoria: no se puede omitir la tercera expresión; la forma completa siempre requiere los tres operandos.

Ejemplo práctico:
int max = (a > b) ? a : b;  // devuelve a si a>b, en caso contrario b.
